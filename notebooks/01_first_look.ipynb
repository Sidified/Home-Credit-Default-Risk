# Home Credit Default Risk — First Look

Explore the application training data before building any model.

Main questions:
- What does the dataset look like?
- How imbalanced is the target?
- How much missing data is there?
- Which features need special attention?
- What should the first baseline look like?

```python
import pandas as pd
import numpy as np

app = pd.read_csv("../data/raw/application_train.csv")

print("Shape:", app.shape)
```

```python
print("Target counts:")
print(app["TARGET"].value_counts())

print("\nTarget proportions:")
print(app["TARGET"].value_counts(normalize=True))
```

```python
target_summary = pd.DataFrame({
    "count": app["TARGET"].value_counts(),
    "percentage": app["TARGET"].value_counts(normalize=True) * 100
})

target_summary
```

```python
print("Data types:")
print(app.dtypes.value_counts())

print("\nNumeric columns:", app.select_dtypes(include=np.number).shape[1])
print("Categorical columns:", app.select_dtypes(include=["object", "str"]).shape[1])
```

```python
missing = (
    app.isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("Columns with missing values:", (missing > 0).sum())
print("Columns with >50% missing:", (missing > 50).sum())

missing.head(20)
```
